```python
from astropy.io import fits
import numpy as np
import matplotlib.pyplot as plt
from photutils.centroids import centroid_2dg, centroid_sources
import os

base_path = "/content/drive/MyDrive/FRD"

files = [
    "F1.fit",
    "F2.fit",
    "F4.fit",
    "F5.fit",
    "F8.fit",
    "f11.fit",
    "f16.fit"
]

box_size_nominal = 91

for fname in files:
    path = os.path.join(base_path, fname)
    image = fits.getdata(path)

    ny, nx = image.shape

    y0, x0 = np.unravel_index(np.nanargmax(image), image.shape)

    half_nominal = box_size_nominal // 2
    half_safe = min(
        half_nominal,
        x0,
        nx - x0 - 1,
        y0,
        ny - y0 - 1
    )
    half_safe = max(5, half_safe)
    box_size = 2 * half_safe + 1

    try:
        x_cen, y_cen = centroid_sources(
            image,
            x0,
            y0,
            box_size=box_size,
            centroid_func=centroid_2dg
        )

        x_cen = float(x_cen)
        y_cen = float(y_cen)
        status = "OK"

    except Exception:
        x_cen, y_cen = np.nan, np.nan
        status = "FAILED"

    print(
        f"{fname} | "
        f"init=({x0},{y0}) | "
        f"box={box_size}px | "
        f"centroid=({x_cen:.2f},{y_cen:.2f}) | {status}"
    )

    plt.figure(figsize=(6,6))
    plt.imshow(
        image,
        origin="lower",
        cmap="gray",
        vmin=np.percentile(image, 5),
        vmax=np.percentile(image, 99)
    )

    plt.plot(x0, y0, "b+", markersize=10, label="Brightest pixel")

    if not np.isnan(x_cen):
        plt.plot(x_cen, y_cen, "r+", markersize=14, mew=2, label="Centroid")

    plt.legend()
    plt.title(fname)
    plt.colorbar(label="Counts (ADU)")
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from scipy.optimize import curve_fit

def gaussian_1d(x, amp, x0, sigma, offset):
    return offset + amp * np.exp(-(x - x0)**2 / (2 * sigma**2))

files = [
    "F1.fit",
    "F2.fit",
    "F4.fit",
    "F5.fit",
    "F8.fit",
    "f11.fit",
    "f16.fit",
    "F22.fit"
]

base_path = "/content/drive/MyDrive/FRD/"

centroids = {
    "F1.fit":  (300, 330),
    "F2.fit":  (298, 332),
    "F4.fit":  (301, 329),
    "F5.fit":  (299, 331),
    "F8.fit":  (302, 328),
    "f11.fit": (310, 345),
    "f16.fit": (315, 350),
    "F22.fit": (320, 360),
}

fig, axes = plt.subplots(len(files), 1, figsize=(8, 3 * len(files)), sharex=False)

for ax, fname in zip(axes, files):
    image = fits.getdata(base_path + fname)

    x0, y0 = centroids[fname]
    y0 = int(y0)

    profile = image[:, y0]
    x = np.arange(len(profile))

    amp0 = profile.max() - np.median(profile)
    x00 = np.argmax(profile)
    sigma0 = 10
    offset0 = np.median(profile)

    popt, _ = curve_fit(
        gaussian_1d, x, profile,
        p0=[amp0, x00, sigma0, offset0],
        maxfev=10000
    )

    sigma = popt[2]
    fwhm = 2.355 * sigma

    x_fine = np.linspace(x.min(), x.max(), 1000)

    ax.plot(x, profile, 'k.', label="Data")
    ax.plot(x_fine, gaussian_1d(x_fine, *popt), 'r-', label="Gaussian fit")
    ax.set_title(f"{fname} | FWHM = {fwhm:.2f} px")
    ax.set_ylabel("Counts")
    ax.legend()

axes[-1].set_xlabel("Pixel index")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from photutils.aperture import CircularAperture, CircularAnnulus, aperture_photometry
from photutils.aperture import ApertureStats

x0, y0 = int(np.round(x)), int(np.round(y))   # centroid
radii = np.arange(2, 130, 2)

annulus = CircularAnnulus((x0, y0), r_in=110, r_out=140)
ann_stats = ApertureStats(image, annulus)
bkg_median = ann_stats.median

fluxes = []

for r in radii:
    ap = CircularAperture((x0, y0), r=r)
    phot = aperture_photometry(image, ap)

    aperture_sum = phot['aperture_sum'][0]
    bkg_total = bkg_median * ap.area

    flux_bkgsub = aperture_sum - bkg_total
    fluxes.append(flux_bkgsub)

fluxes = np.array(fluxes)

plt.figure(figsize=(6,4))
plt.plot(radii, fluxes, 'ko-', lw=1)
plt.xlabel('Aperture Radius (pixels)')
plt.ylabel('Background-subtracted Flux (ADU)')
plt.title('Growth Curve')
plt.grid(alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from photutils.aperture import CircularAperture, CircularAnnulus, aperture_photometry
from photutils.aperture import ApertureStats

files = [
    "F1.fit", "F2.fit", "F4.fit", "F5.fit",
    "F8.fit", "f11.fit", "f16.fit", "F22.fit"
]

base_path = "/content/drive/MyDrive/FRD/"

centroids = {
    "F1.fit":  (300, 330),
    "F2.fit":  (298, 332),
    "F4.fit":  (301, 329),
    "F5.fit":  (299, 331),
    "F8.fit":  (302, 328),
    "f11.fit": (310, 345),
    "f16.fit": (315, 350),
    "F22.fit": (320, 360),
}

radii = np.arange(2, 130, 2)

all_fluxes = {}

for fname in files:
    image = fits.getdata(base_path + fname)
    x0, y0 = map(int, centroids[fname])

    annulus = CircularAnnulus((x0, y0), r_in=110, r_out=140)
    ann_stats = ApertureStats(image, annulus)
    bkg_median = ann_stats.median

    image_bs = image - bkg_median

    plt.figure(figsize=(5,5))
    plt.imshow(
        image_bs,
        origin='lower',
        cmap='gray',
        vmin=np.percentile(image_bs, 5),
        vmax=np.percentile(image_bs, 99)
    )
    plt.colorbar(label='Counts (ADU)')
    plt.title(f"{fname} (background subtracted)")
    plt.show()

    fluxes = []
    for r in radii:
        ap = CircularAperture((x0, y0), r=r)
        phot = aperture_photometry(image_bs, ap)
        fluxes.append(phot['aperture_sum'][0])

    all_fluxes[fname] = np.array(fluxes)

plt.figure(figsize=(8,6))
for fname, fluxes in all_fluxes.items():
    plt.plot(radii, fluxes, label=fname)

plt.xlabel("Aperture Radius (pixels)")
plt.ylabel("Background-subtracted Flux (ADU)")
plt.title("Growth Curves Comparison (All Images)")
plt.legend(fontsize=8)
plt.grid(alpha=0.3)
plt.show()
```
